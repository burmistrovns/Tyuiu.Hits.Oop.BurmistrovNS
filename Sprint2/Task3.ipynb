<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Book в C#, который будет представлять информацию о
книгах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Book:
• Атрибуты: Название (Title), Автор (Author), Год издания (YearOfPublication).
• Методы:
o GetInfo(): метод для получения информации о книге в виде строки.
o Read(): метод для вывода сообщения о чтении книги.
o Borrow(): метод для вывода сообщения о выдаче книги на чтение.

Требования к производным классам:
1. Учебник (Textbook): Должен содержать дополнительные атрибуты, такие как
Предмет (Subject). Метод Read() должен быть переопределен для
добавления информации о предмете при чтении учебника.
2. Художественная литература (Fiction): Должен содержать дополнительные
атрибуты, такие как Жанр (Genre). Метод Borrow() должен быть
переопределен для добавления информации о жанре при выдаче книги на
чтение.
3. Научная литература (ScientificLiterature) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Область науки
(FieldOfScience). Метод GetInfo() должен быть переопределен для включения
информации об области науки в описании книги.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
using System;
using System.Collections.Generic;

public interface IDownloadable
{
    double FileSize { get; set; }
    void Download();
}

public interface IPlayable
{
    double Duration { get; set; }
    void Play();
}

public class Book
{
    private string _title;
    private string _author;
    private int _year;
    private bool _isBorrowed;

    public string Title
    {
        get { return _title; }
        set { _title = string.IsNullOrEmpty(value) ? "Без названия" : value; }
    }

    public string Author
    {
        get { return _author; }
        set { _author = string.IsNullOrEmpty(value) ? "Неизвестный автор" : value; }
    }

    public int YearOfPublication
    {
        get { return _year; }
        set { _year = (value >= 0 && value <= 2026) ? value : 2000; }
    }

    public bool IsBorrowed
    {
        get { return _isBorrowed; }
        set { _isBorrowed = value; }
    }

    public string ISBN { get; set; }
    public string Publisher { get; set; }
    public int PageCount { get; set; }
    public string Language { get; set; }

    public Book(string title, string author, int yearOfPublication, string isbn)
    {
        Title = title;
        Author = author;
        YearOfPublication = yearOfPublication;
        ISBN = isbn;
        IsBorrowed = false;
        Language = "Русский";
    }

    public virtual string GetInfo()
    {
        return $"«{Title}» — {Author} ({YearOfPublication} г., ISBN: {ISBN})";
    }

    public virtual void Read()
    {
        Console.WriteLine($"Вы читаете книгу: «{Title}» автора {Author}");
    }

    public virtual void Borrow()
    {
        if (!IsBorrowed)
        {
            IsBorrowed = true;
            Console.WriteLine($"Книга «{Title}» успешно выдана на чтение");
        }
        else
        {
            Console.WriteLine($"Книга «{Title}» уже кем-то занята");
        }
    }

    public void OpenBook()
    {
        Console.WriteLine($"Книга «{Title}» открыта.");
    }

    public void CloseBook()
    {
        Console.WriteLine($"Книга «{Title}» закрыта.");
    }

    public void SetBookmark(int page)
    {
        if (page > 0 && page <= PageCount)
            Console.WriteLine($"Закладка установлена на {page} странице.");
        else
            Console.WriteLine("Неверный номер страницы для закладки.");
    }

    public int GetBookAge()
    {
        return DateTime.Now.Year - YearOfPublication;
    }
}

public class Textbook : Book
{
    public string Subject { get; set; }
    public int TargetGrade { get; set; }
    public bool HasExercises { get; set; }
    public string EducationalStandard { get; set; }

    public Textbook(string title, string author, int year, string isbn, string subject, int grade) 
        : base(title, author, year, isbn)
    {
        Subject = subject;
        TargetGrade = grade;
    }

    public override void Read()
    {
        Console.WriteLine($"Вы изучаете учебник по предмету \"{Subject}\" для {TargetGrade} класса: «{Title}»");
    }

    public void SolveTask(int taskNumber)
    {
        if (HasExercises)
            Console.WriteLine($"Решение задачи №{taskNumber} из учебника «{Title}».");
        else
            Console.WriteLine("В этом учебнике нет практических задач.");
    }

    public void ShowAnswers()
    {
        Console.WriteLine("Ответы к учебнику открыты.");
    }

    public string GetDifficultyLevel()
    {
        return TargetGrade > 9 ? "Сложный" : "Базовый";
    }
}

public class Fiction : Book
{
    public string Genre { get; set; }
    public string MainCharacter { get; set; }
    public string Setting { get; set; }
    public int AgeRestriction { get; set; }

    public Fiction(string title, string author, int year, string isbn, string genre, int ageRestriction) 
        : base(title, author, year, isbn)
    {
        Genre = genre;
        AgeRestriction = ageRestriction;
    }

    public override void Borrow()
    {
        if (!IsBorrowed)
        {
            IsBorrowed = true;
            Console.WriteLine($"Выдана художественная книга «{Title}» (Жанр: {Genre}, Возраст: {AgeRestriction}+)");
        }
        else
        {
            Console.WriteLine($"Роман «{Title}» сейчас недоступен");
        }
    }

    public void GetSummary()
    {
        Console.WriteLine($"Краткое содержание: Главный герой {MainCharacter} в сеттинге {Setting}.");
    }

    public bool CheckAge(int userAge)
    {
        return userAge >= AgeRestriction;
    }

    public void RecommendSimilar()
    {
        Console.WriteLine($"Рекомендуем другие книги в жанре {Genre}.");
    }
}

public class ScientificLiterature : Book
{
    public string FieldOfScience { get; set; }
    public double ImpactFactor { get; set; }
    public bool IsPeerReviewed { get; set; }
    public string ResearchInstitution { get; set; }

    public ScientificLiterature(string title, string author, int year, string isbn, string field) 
        : base(title, author, year, isbn)
    {
        FieldOfScience = field;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $" [Научная область: {FieldOfScience}, Импакт-фактор: {ImpactFactor}]";
    }

    public void Cite()
    {
        Console.WriteLine($"Цитата: {Author} ({YearOfPublication}). {Title}. {ResearchInstitution}.");
    }

    public void ViewReferences()
    {
        Console.WriteLine($"Открыт список литературы для «{Title}».");
    }

    public void ValidateHypothesis()
    {
        Console.WriteLine(IsPeerReviewed ? "Гипотеза подтверждена научным сообществом." : "Требуется рецензирование.");
    }
}

public class AudioFiction : Fiction, IDownloadable, IPlayable
{
    public double FileSize { get; set; }
    public double Duration { get; set; }
    public string Narrator { get; set; }
    public string AudioFormat { get; set; }

    public AudioFiction(string title, string author, int year, string isbn, string genre, int age, string narrator) 
        : base(title, author, year, isbn, genre, age)
    {
        Narrator = narrator;
    }

    public void Download()
    {
        Console.WriteLine($"Скачивание аудиокниги «{Title}». Размер: {FileSize} МБ.");
    }

    public void Play()
    {
        Console.WriteLine($"Воспроизведение аудиокниги «{Title}». Читает {Narrator}. Длительность: {Duration} часов.");
    }

    public void RateNarrator(int rating)
    {
        Console.WriteLine($"Оценка диктора {Narrator}: {rating}/10");
    }
}

public class Reader
{
    public string Name { get; set; }
    public int ReaderId { get; set; }
    public string Email { get; set; }
    public int MaxBooksAllowed { get; set; }
    
    private List<Book> _currentBooks;

    public Reader(string name, int id)
    {
        Name = name;
        ReaderId = id;
        MaxBooksAllowed = 5;
        _currentBooks = new List<Book>();
    }

    public void TakeBook(Book book)
    {
        Console.WriteLine($"Читатель {Name} пытается взять книгу \"{book.Title}\"");
        
        if (book is Fiction fictionBook && !fictionBook.CheckAge(16)) 
        {
            Console.WriteLine($"{Name} не проходит по возрастному ограничению для этой книги.");
            return;
        }

        if (_currentBooks.Count >= MaxBooksAllowed)
        {
            Console.WriteLine($"{Name} превысил лимит книг.");
            return;
        }

        if (!book.IsBorrowed)
        {
            book.Borrow();
            _currentBooks.Add(book);
        }
        else
        {
            Console.WriteLine($"К сожалению, {Name} не может взять эту книгу");
        }
    }

    public void ReturnBook(Book book)
    {
        if (_currentBooks.Contains(book))
        {
            book.IsBorrowed = false;
            _currentBooks.Remove(book);
            Console.WriteLine($"{Name} вернул книгу «{book.Title}».");
        }
    }

    public void ReadBooks()
    {
        if (_currentBooks.Count == 0)
        {
            Console.WriteLine($"{Name} сейчас ничего не читает");
            return;
        }

        Console.WriteLine($"{Name} начинает читать свои книги:");
        foreach (var book in _currentBooks)
        {
            book.Read();
            book.SetBookmark(15);
        }
    }

    public void ShowProfile()
    {
        Console.WriteLine($"Профиль читателя: {Name} (ID: {ReaderId}), Взято книг: {_currentBooks.Count}/{MaxBooksAllowed}");
    }

    public void SendNotification(string msg)
    {
        Console.WriteLine($"Уведомление для {Name} ({Email}): {msg}");
    }
}

Textbook mathBook = new Textbook("Алгебра", "Колмогоров", 2019, "978-5-09", "Математика", 10) 
{ 
    HasExercises = true, 
    PageCount = 350 
};

Fiction fantasyBook = new Fiction("Властелин Колец", "Толкин", 1954, "978-5-17", "Фэнтези", 12) 
{ 
    MainCharacter = "Фродо", 
    Setting = "Средиземье", 
    PageCount = 1000 
};

ScientificLiterature spaceBook = new ScientificLiterature("Краткая история времени", "Хокинг", 1988, "978-5-38", "Астрофизика") 
{ 
    ImpactFactor = 8.5, 
    IsPeerReviewed = true 
};

AudioFiction audioWitcher = new AudioFiction("Ведьмак", "Сапковский", 1986, "978-5-11", "Фэнтези", 18, "Всеволод Кузнецов")
{
    FileSize = 450.5,
    Duration = 12.5,
    AudioFormat = "MP3"
};

Reader reader1 = new Reader("Иванов Иван", 101) { Email = "ivan@test.com" };
Reader reader2 = new Reader("Петров Петр", 102) { Email = "petr@test.com" };

Console.WriteLine("--- ДЕМОНСТРАЦИЯ НАСЛЕДОВАНИЯ И ВЗАИМОДЕЙСТВИЯ ---\n");

reader1.TakeBook(fantasyBook);
reader1.TakeBook(mathBook);
reader2.TakeBook(fantasyBook); 

Console.WriteLine();
reader1.ReadBooks();
mathBook.SolveTask(42);
fantasyBook.GetSummary();

Console.WriteLine("\n--- ДЕМОНСТРАЦИЯ СЛОЖНОГО И МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ ---");
audioWitcher.Download();
audioWitcher.Play();
audioWitcher.RateNarrator(10);
audioWitcher.CheckAge(20); 

Console.WriteLine("\n--- ПРОФИЛИ ---");
reader1.ShowProfile();
reader1.ReturnBook(mathBook);
reader1.ShowProfile();
```
