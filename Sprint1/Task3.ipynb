### Вариант задания 5


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Book в C#, который будет представлять информацию о
книгах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Book:
• Атрибуты: Название (Title), Автор (Author), Год издания (YearOfPublication).
• Методы:
o GetInfo(): метод для получения информации о книге в виде строки.
o Read(): метод для вывода сообщения о чтении книги.
o Borrow(): метод для вывода сообщения о выдаче книги на чтение.

Требования к производным классам:
1. Учебник (Textbook): Должен содержать дополнительные атрибуты, такие как
Предмет (Subject). Метод Read() должен быть переопределен для
добавления информации о предмете при чтении учебника.
2. Художественная литература (Fiction): Должен содержать дополнительные
атрибуты, такие как Жанр (Genre). Метод Borrow() должен быть
переопределен для добавления информации о жанре при выдаче книги на
чтение.
3. Научная литература (ScientificLiterature) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Область науки
(FieldOfScience). Метод GetInfo() должен быть переопределен для включения
информации об области науки в описании книги.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

public class Book
{
    private string _title;
    private string _author;
    private int _year;
    private bool _isBorrowed;

    public string Title
    {
        get { return _title; }
        set 
        { 
            if (!string.IsNullOrEmpty(value)) _title = value; 
            else _title = "Без названия";
        }
    }

    public string Author
    {
        get { return _author; }
        set 
        { 
            if (!string.IsNullOrEmpty(value)) _author = value; 
            else _author = "Неизвестный автор";
        }
    }

    public int YearOfPublication
    {
        get { return _year; }
        set 
        { 
            if (value >= 0 && value <= 2026) _year = value; 
            else _year = 2000;
        }
    }

    public bool IsBorrowed
    {
        get { return _isBorrowed; }
        set { _isBorrowed = value; }
    }

    public Book(string title, string author, int yearOfPublication)
    {
        Title = title;
        Author = author;
        YearOfPublication = yearOfPublication;
        IsBorrowed = false;
    }

    public virtual string GetInfo()
    {
        return $"«{Title}» — {Author} ({YearOfPublication} г.)";
    }

    public virtual void Read()
    {
        Console.WriteLine($"Вы читаете книгу: «{Title}» автора {Author}.");
    }

    public virtual void Borrow()
    {
        if (!_isBorrowed)
        {
            _isBorrowed = true;
            Console.WriteLine($"Книга «{Title}» успешно выдана на чтение.");
        }
        else
        {
            Console.WriteLine($"Книга «{Title}» уже кем-то занята!");
        }
    }
}

public class Textbook : Book
{
    private string _subject;

    public string Subject
    {
        get { return _subject; }
        set { _subject = value; }
    }

    public Textbook(string title, string author, int yearOfPublication, string subject) 
        : base(title, author, yearOfPublication)
    {
        Subject = subject;
    }

    public override void Read()
    {
        Console.WriteLine($"Вы изучаете учебник по предмету \"{Subject}\": «{Title}» ({Author}).");
    }
}

public class Fiction : Book
{
    private string _genre;

    public string Genre
    {
        get { return _genre; }
        set { _genre = value; }
    }

    public Fiction(string title, string author, int yearOfPublication, string genre) 
        : base(title, author, yearOfPublication)
    {
        Genre = genre;
    }

    public override void Borrow()
    {
        if (!IsBorrowed)
        {
            IsBorrowed = true;
            Console.WriteLine($"Выдана художественная книга «{Title}» (Жанр: {Genre}). Приятного чтения!");
        }
        else
        {
            Console.WriteLine($"Роман «{Title}» сейчас недоступен.");
        }
    }
}

public class ScientificLiterature : Book
{
    private string _fieldOfScience;

    public string FieldOfScience
    {
        get { return _fieldOfScience; }
        set { _fieldOfScience = value; }
    }

    public ScientificLiterature(string title, string author, int yearOfPublication, string fieldOfScience) 
        : base(title, author, yearOfPublication)
    {
        FieldOfScience = fieldOfScience;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $" [Научная область: {FieldOfScience}]";
    }
}

public class Reader
{
    private string _name;
    private List<Book> _currentBooks;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public Reader(string name)
    {
        Name = name;
        _currentBooks = new List<Book>();
    }

    public void TakeBook(Book book)
    {
        Console.WriteLine($"Читатель {Name} пытается взять книгу \"{book.Title}\"...");
        if (!book.IsBorrowed)
        {
            book.Borrow();
            _currentBooks.Add(book);
        }
        else
        {
            Console.WriteLine($"К сожалению, {Name} не может взять эту книгу.");
        }
    }

    public void ReadBooks()
    {
        if (_currentBooks.Count == 0)
        {
            Console.WriteLine($"{Name} сейчас ничего не читает.");
            return;
        }

        Console.WriteLine($"---{Name} начинает читать свои книги:");
        foreach (var book in _currentBooks)
        {
            book.Read();
        }
    }
}

List<Book> library = new List<Book>
{
    new Book("Чистый код", "Роберт Мартин", 2008),
    new Textbook("Алгебра и начала анализа", "Колмогоров", 2019, "Математика"),
    new Fiction("Властелин Колец", "Джон Толкин", 1954, "Фэнтези"),
    new ScientificLiterature("Краткая история времени", "Стивен Хокинг", 1988, "Астрофизика")
};

Reader reader1 = new Reader("Иванов Иван");
Reader reader2 = new Reader("Петров Петр");

Console.WriteLine("Проверка взаимодействия объектов\n");

reader1.TakeBook(library[0]); 
Console.WriteLine();

reader2.TakeBook(library[0]); 
Console.WriteLine();

reader1.TakeBook(library[1]); 
Console.WriteLine();

reader1.ReadBooks();
Console.WriteLine();

reader2.ReadBooks();
Console.WriteLine();

```
