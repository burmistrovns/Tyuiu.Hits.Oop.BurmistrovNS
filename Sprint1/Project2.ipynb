```polyglot-notebook
using System;
using System.Collections.Generic;

public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { if (value >= 0) price = value; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { if (value >= 0) quantity = value; }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    public Product(string name, decimal price, int quantity, string cat, string desc)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = cat;
        this.description = desc;
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        quantity += amount;
    }

    public string GetProductInfo()
    {
        return "Товар: " + name + ", Цена: " + price + ", Кол-во: " + quantity + ", Категория: " + category + ", Описание: " + description;
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;
    
    private List<Product> list = new List<Product>();

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set { if (value >= 0) salary = value; }
    }

    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public List<Product> List
    {
        get { return list; }
    }

    public Seller(string name, string id, decimal salary, string info)
    {
        this.name = name;
        this.employeeId = id;
        this.salary = salary;
        this.contactInfo = info;
    }

    public void AddProduct(Product product)
    {
        list.Add(product);
    }

    public void SellProduct(Product product, int quantity)
    {
        if (product.Quantity >= quantity)
        {
            product.Quantity -= quantity;
        }
    }

    public string GetSellerInfo()
    {
        return "Продавец: " + name + ", ID: " + employeeId + ", Контакты: " + contactInfo;
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;

    private List<Seller> staff = new List<Seller>();

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public string StoreHours
    {
        get { return storeHours; }
        set { storeHours = value; }
    }

    public Store(string name, string loc, string hours)
    {
        this.storeName = name;
        this.location = loc;
        this.storeHours = hours;
    }

    public void AddSeller(Seller seller)
    {
        staff.Add(seller);
    }

    public void ListProducts()
    {
        Console.WriteLine("Товары в магазине " + storeName + ":");
        foreach (var s in staff)
        {
            foreach (var p in s.List)
            {
                Console.WriteLine(p.GetProductInfo());
            }
        }
    }

    public string GetStoreInfo()
    {
        return "Магазин: " + storeName + ", Адрес: " + location + ", Часы: " + storeHours;
    }
}

Product p = new Product("Телефон", 25000m, 10, "Гаджеты", "Черный смартфон");
Seller s = new Seller("Иван", "EMP1", 45000m, "ivan@mail.com");
Store store = new Store("Техно", "ул. Ленина 10", "9:00 - 21:00");

store.AddSeller(s);
s.AddProduct(p);

Console.WriteLine(store.GetStoreInfo());
Console.WriteLine(s.GetSellerInfo());
Console.WriteLine(p.GetProductInfo());
Console.WriteLine("Общая стоимость: " + p.GetTotalPrice());

Console.WriteLine("-------------------------------------------------------------------------------------------------------");
store.ListProducts();

Console.WriteLine("Продажа 3 штук");
s.SellProduct(p, 3);
Console.WriteLine(p.GetProductInfo());

Console.WriteLine("Обновление количества на +5");
p.UpdateQuantity(5);
Console.WriteLine(p.GetProductInfo());
```
