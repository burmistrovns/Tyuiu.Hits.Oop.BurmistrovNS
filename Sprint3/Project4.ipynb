```polyglot-notebook
using System;
using System.Collections.Generic;

public interface IInstallable
{
    void Install();
}

public abstract class Component
{
    public string Name { get; set; }
    public double Price { get; set; }

    public Component(string n, double p)
    {
        Name = n;
        Price = p;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine("Компонент: " + Name + ", Цена: " + Price);
    }

    public override string ToString()
    {
        return Name + " (" + Price + " руб.)";
    }
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }

    public CPU(string n, double p, int c) : base(n, p)
    {
        Cores = c;
    }

    public override void GetInfo()
    {
        Console.WriteLine("Процессор: " + Name + ", Ядра: " + Cores + ", Цена: " + Price);
    }

    public void Install()
    {
        Console.WriteLine("Процессор " + Name + " установлен в сокет.");
    }

    public static bool operator ==(CPU a, CPU b)
    {
        if (object.ReferenceEquals(a, null) || object.ReferenceEquals(b, null)) 
            return object.ReferenceEquals(a, b);
        return a.Name == b.Name && a.Cores == b.Cores;
    }

    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is CPU c) return this == c;
        return false;
    }

    public override int GetHashCode()
    {
        return Name.GetHashCode() + Cores;
    }
}

public class RAM : Component, IInstallable
{
    public int Size { get; set; }

    public RAM(string n, double p, int s) : base(n, p)
    {
        Size = s;
    }

    public override void GetInfo()
    {
        Console.WriteLine("ОЗУ: " + Name + ", Объем: " + Size + "ГБ, Цена: " + Price);
    }

    public void Install()
    {
        Console.WriteLine("Планка памяти " + Name + " вставлена в слот.");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        if (object.ReferenceEquals(a, null) || object.ReferenceEquals(b, null)) 
            return object.ReferenceEquals(a, b);
        return a.Name == b.Name && a.Size == b.Size;
    }

    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is RAM r) return this == r;
        return false;
    }

    public override int GetHashCode()
    {
        return Name.GetHashCode() + Size;
    }
}

public class GPU : Component, IInstallable
{
    public int Vram { get; set; }

    public GPU(string n, double p, int v) : base(n, p)
    {
        Vram = v;
    }

    public override void GetInfo()
    {
        Console.WriteLine("Видеокарта: " + Name + ", Память: " + Vram + "ГБ, Цена: " + Price);
    }

    public void Install()
    {
        Console.WriteLine("Видеокарта " + Name + " подключена к материнской плате.");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        if (object.ReferenceEquals(a, null) || object.ReferenceEquals(b, null)) 
            return object.ReferenceEquals(a, b);
        return a.Name == b.Name && a.Vram == b.Vram;
    }

    public static bool operator !=(GPU a, GPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is GPU g) return this == g;
        return false;
    }

    public override int GetHashCode()
    {
        return Name.GetHashCode() + Vram;
    }
}

public class HDD : Component, IInstallable
{
    public int Capacity { get; set; }

    public HDD(string n, double p, int c) : base(n, p)
    {
        Capacity = c;
    }

    public override void GetInfo()
    {
        Console.WriteLine("Жесткий диск: " + Name + ", Объем: " + Capacity + "ГБ, Цена: " + Price);
    }

    public void Install()
    {
        Console.WriteLine("Жесткий диск " + Name + " закреплен в корпусе.");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        if (object.ReferenceEquals(a, null) || object.ReferenceEquals(b, null)) 
            return object.ReferenceEquals(a, b);
        return a.Name == b.Name && a.Capacity == b.Capacity;
    }

    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        if (obj is HDD h) return this == h;
        return false;
    }

    public override int GetHashCode()
    {
        return Name.GetHashCode() + Capacity;
    }
}

public class Computer
{
    private List<Component> components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        component.Install();
        components.Add(component);
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        components.Remove(component);
        Console.WriteLine("Удалено: " + component.Name);
    }

    public double GetTotalPrice()
    {
        double total = 0;
        for (int i = 0; i < components.Count; i++)
        {
            total = total + components[i].Price;
        }
        return total;
    }

    public void ShowAllComponents()
    {
        for (int i = 0; i < components.Count; i++)
        {
            components[i].GetInfo();
        }
    }
}

public static class ComputerFactory
{
    public static Computer BuildPC<T1, T2, T3, T4>(T1 c1, T2 c2, T3 c3, T4 c4) 
        where T1 : CPU, IInstallable
        where T2 : RAM, IInstallable
        where T3 : GPU, IInstallable
        where T4 : HDD, IInstallable
    {
        Computer pc = new Computer();
        pc.AddComponent(c1);
        pc.AddComponent(c2);
        pc.AddComponent(c3);
        pc.AddComponent(c4);
        return pc;
    }
}

CPU myCpu = new CPU("Ryzen 5 3600", 4000, 6);
CPU anotherCpu = new CPU("Ryzen 5 3600", 4000, 6);
RAM myRam = new RAM("Netac", 7000, 16);
GPU myGpu = new GPU("RTX 2070 Super", 20000, 8);
HDD myHdd = new HDD("Kingston", 5000, 1000);

Console.WriteLine("Сборка ПК через фабрику:");
Computer myPC = ComputerFactory.BuildPC(myCpu, myRam, myGpu, myHdd);

Console.WriteLine();
Console.WriteLine("Свойства собранного ПК:");
myPC.ShowAllComponents();

Console.WriteLine();
Console.WriteLine("Общая стоимость: " + myPC.GetTotalPrice() + " руб.");

Console.WriteLine();
Console.WriteLine("Проверка работы оператора ToString:");
Console.WriteLine(myCpu.ToString());

Console.WriteLine();
Console.WriteLine("Проверка оператора сравнения (==):");
if (myCpu == anotherCpu)
{
    Console.WriteLine("Процессоры одинаковые.");
}
else
{
    Console.WriteLine("Процессоры разные.");
}

Console.WriteLine();
myPC.RemoveComponent(myHdd);
Console.WriteLine("Новая стоимость: " + myPC.GetTotalPrice() + " руб.");
```
