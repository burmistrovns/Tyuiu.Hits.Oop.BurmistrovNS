### Вариант задания 5


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Book в C#, который будет представлять информацию о
книгах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Book:
• Атрибуты: Название (Title), Автор (Author), Год издания (YearOfPublication).
• Методы:
o GetInfo(): метод для получения информации о книге в виде строки.
o Read(): метод для вывода сообщения о чтении книги.
o Borrow(): метод для вывода сообщения о выдаче книги на чтение.

Требования к производным классам:
1. Учебник (Textbook): Должен содержать дополнительные атрибуты, такие как
Предмет (Subject). Метод Read() должен быть переопределен для
добавления информации о предмете при чтении учебника.
2. Художественная литература (Fiction): Должен содержать дополнительные
атрибуты, такие как Жанр (Genre). Метод Borrow() должен быть
переопределен для добавления информации о жанре при выдаче книги на
чтение.
3. Научная литература (ScientificLiterature) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Область науки
(FieldOfScience). Метод GetInfo() должен быть переопределен для включения
информации об области науки в описании книги.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
using System;
using System.Collections.Generic;

public interface IDownloadable
{
    double FileSize { get; set; }
    void Download();
}

public interface IPlayable
{
    double Duration { get; set; }
    void Play();
}

public class Book
{
    private string _title;
    private string _author;
    private int _year;

    public string Title
    {
        get { return _title; }
        set { _title = string.IsNullOrEmpty(value) ? "Без названия" : value; }
    }

    public string Author
    {
        get { return _author; }
        set { _author = string.IsNullOrEmpty(value) ? "Неизвестный автор" : value; }
    }

    public int YearOfPublication
    {
        get { return _year; }
        set { _year = (value > 0) ? value : 2000; }
    }

    public bool IsBorrowed { get; set; }
    public string ISBN { get; set; }

    public Book(string title, string author, int year, string isbn)
    {
        Title = title;
        Author = author;
        YearOfPublication = year;
        ISBN = isbn;
        IsBorrowed = false;
    }

    public virtual string GetInfo()
    {
        return $"«{Title}» — {Author} ({YearOfPublication} г.)";
    }

    public string GetInfo(bool includeIsbn)
    {
        if (includeIsbn)
            return $"«{Title}» — {Author} ({YearOfPublication} г.), ISBN: {ISBN}";
        return GetInfo();
    }

    public virtual void Read()
    {
        Console.WriteLine($"Вы читаете книгу: «{Title}»");
    }

    public void Read(int hours)
    {
        Console.WriteLine($"Вы читали книгу «{Title}» на протяжении {hours} часов.");
    }

    public void Read(string location)
    {
        Console.WriteLine($"Вы читаете «{Title}» в локации: {location}.");
    }

    public virtual void Borrow()
    {
        if (!IsBorrowed)
        {
            IsBorrowed = true;
            Console.WriteLine($"Книга «{Title}» выдана.");
        }
        else
        {
            Console.WriteLine($"Книга «{Title}» занята.");
        }
    }
}

public class Textbook : Book
{
    public string Subject { get; set; }
    public int TargetGrade { get; set; }
    public bool HasExercises { get; set; }

    public Textbook(string title, string author, int year, string isbn, string subject, int grade) 
        : base(title, author, year, isbn)
    {
        Subject = subject;
        TargetGrade = grade;
        HasExercises = true;
    }

    public override void Read()
    {
        Console.WriteLine($"Изучение учебника по предмету \"{Subject}\" для {TargetGrade} класса: «{Title}»");
    }

    public void SolveTask()
    {
        Console.WriteLine("Решение случайной задачи из учебника.");
    }

    public void SolveTask(int taskNumber)
    {
        Console.WriteLine($"Решение задачи №{taskNumber} из учебника «{Title}».");
    }
}

public class Fiction : Book
{
    public string Genre { get; set; }
    public int AgeRestriction { get; set; }
    public string MainCharacter { get; set; }

    public Fiction(string title, string author, int year, string isbn, string genre, int ageRestriction) 
        : base(title, author, year, isbn)
    {
        Genre = genre;
        AgeRestriction = ageRestriction;
    }

    public override void Borrow()
    {
        if (!IsBorrowed)
        {
            IsBorrowed = true;
            Console.WriteLine($"Выдана художественная книга «{Title}» (Жанр: {Genre}).");
        }
        else
        {
            Console.WriteLine($"Роман «{Title}» недоступен.");
        }
    }

    public bool CheckAge(int userAge)
    {
        return userAge >= AgeRestriction;
    }

    public bool CheckAge(int userAge, bool hasParentalConsent)
    {
        if (hasParentalConsent) return true;
        return userAge >= AgeRestriction;
    }
}

public class ScientificLiterature : Book
{
    public string FieldOfScience { get; set; }
    public double ImpactFactor { get; set; }
    public bool IsPeerReviewed { get; set; }

    public ScientificLiterature(string title, string author, int year, string isbn, string field) 
        : base(title, author, year, isbn)
    {
        FieldOfScience = field;
        IsPeerReviewed = true;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $" [Область: {FieldOfScience}]";
    }
}

public class AudioFiction : Fiction, IDownloadable, IPlayable
{
    public double FileSize { get; set; }
    public double Duration { get; set; }
    public string Narrator { get; set; }

    public AudioFiction(string title, string author, int year, string isbn, string genre, int age, string narrator) 
        : base(title, author, year, isbn, genre, age)
    {
        Narrator = narrator;
    }

    public void Download()
    {
        Console.WriteLine($"Скачивание аудиокниги «{Title}» ({FileSize} МБ).");
    }

    public void Play()
    {
        Console.WriteLine($"Воспроизведение аудиокниги «{Title}». Читает {Narrator}.");
    }
}

public class LibraryCatalog<T> where T : Book
{
    private List<T> _items = new List<T>();
    public string CatalogName { get; set; }

    public LibraryCatalog(string name)
    {
        CatalogName = name;
    }

    public void AddItem(T item)
    {
        _items.Add(item);
    }

    public void RemoveItem(T item)
    {
        _items.Remove(item);
    }

    public void ShowAllItems()
    {
        Console.WriteLine($"\n--- Каталог: {CatalogName} ---");
        foreach (var item in _items)
        {
            Console.WriteLine(item.GetInfo(true));
        }
    }
    
    public void ReadAll()
    {
        foreach (var item in _items)
        {
            item.Read();
        }
    }
}

Textbook mathBook = new Textbook("Алгебра", "Колмогоров", 2019, "978-5-09", "Математика", 10);
Fiction fantasyBook = new Fiction("Властелин Колец", "Толкин", 1954, "978-5-17", "Фэнтези", 12);
ScientificLiterature spaceBook = new ScientificLiterature("Краткая история времени", "Хокинг", 1988, "978-5-38", "Астрофизика");
AudioFiction audioBook = new AudioFiction("Ведьмак", "Сапковский", 1986, "978-5-11", "Фэнтези", 18, "Кузнецов") { FileSize = 450.5 };

LibraryCatalog<Book> generalLibrary = new LibraryCatalog<Book>("Общая библиотека");
generalLibrary.AddItem(mathBook);
generalLibrary.AddItem(fantasyBook);
generalLibrary.AddItem(spaceBook);
generalLibrary.AddItem(audioBook);

generalLibrary.ShowAllItems();

Console.WriteLine("\n--- Демонстрация перекрытия и перегрузки ---");
mathBook.Read(); 
fantasyBook.Read(3); 
spaceBook.Read("Библиотека"); 

Console.WriteLine();
mathBook.SolveTask();
mathBook.SolveTask(42); 

Console.WriteLine();
Console.WriteLine($"Проверка возраста (15 лет): {fantasyBook.CheckAge(15)}");
Console.WriteLine($"Проверка возраста (10 лет с разрешением): {fantasyBook.CheckAge(10, true)}");
```
