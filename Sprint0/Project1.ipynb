```polyglot-notebook
using System;

public class Plant
{
    private string _name;
    private string _type;
    private float _height;

    public string Name { get { return _name; } set { _name = value; } }
    public string Type { get { return _type; } set { _type = value; } }
    public float Height { get { return _height; } set { _height = Math.Max(0, value); } }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветет.");
    }

    public void Grow(float h)
    {
        Height += h;
        Console.WriteLine($"{Name} выросло на {h} см. Теперь: {Height} см.");
    }
}

public class Flower : Plant
{
    private string _color;
    private int _petals;

    public string Color { get { return _color; } set { _color = value; } }
    public int Petals { get { return _petals; } set { _petals = Math.Max(0, value); } }

    public Flower(string name, float height, string color, int petals) 
        : base(name, "Цветок", height)
    {
        Color = color;
        Petals = petals;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void Wither()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать...");
    }
}

public class Tree : Plant
{
    private int _age;
    private string _leaf;

    public int Age { get { return _age; } set { _age = Math.Max(0, value); } }
    public string Leaf { get { return _leaf; } set { _leaf = value; } }

    public Tree(string name, float height, int age, string leaf) 
        : base(name, "Дерево", height)
    {
        Age = age;
        Leaf = leaf;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void GrowYears(int y)
    {
        Age += y;
        Console.WriteLine($"Прошло {y} л. Новый возраст {Name}: {Age} л.");
    }
}

Flower rose = new Flower("Роза", 45.5f, "Красный", 15);
Tree oak = new Tree("Дуб", 320f, 12, "Лиственные");

rose.Grow(5.5f);
oak.Grow(15f);
oak.GrowYears(3);

Console.WriteLine("---");

Plant p1 = rose;
Plant p2 = oak;

p1.Bloom();
p2.Bloom();

Console.WriteLine("---");

rose.Wither();
```
